### SECTION 5: First simple AI call
The OpenAI chat API expects a list of messages. Each message normally has:
 - role: who is speaking
 - content: the actual message

Here, the user asks a simple question.

Why is it a list?

Because later you can have several messages, like a conversation:

messages = [
    {"role": "system", "content": "You are a helpful tutor."},
    {"role": "user", "content": "What is 2+2?"},
    {"role": "assistant", "content": "2+2 is 4."},
    {"role": "user", "content": "Why?"}
]

```python
messages = [
    {
        "role": "user",
        "content": "What is 2+2?"
    }
]
```

The following section sends the message to the model.

gpt-4.1-nano is a small model, useful for simple/cheap tests.

What gets stored in response?

This part:

`response =`

stores the model’s answer inside a variable named response.

The answer is not just plain text. It is an object with several pieces of information, such as:

the model response text,
model used,
token usage,
metadata.

Then this part:

`openai.chat.completions.create(...)`

means:

`Create a chat completion using the OpenAI client.`

```python
response = openai.chat.completions.create(
    model="gpt-4.1-nano",
    messages=messages
)
```

The following section prints the actual answer from the AI.

`response`

This is the object returned by the OpenAI API call:

`response = openai.chat.completions.create(...)`

It contains many things, not only the answer text. It may contain metadata such as model name, response ID, token usage, and the actual response message.

`response.choices`

This means:

Go inside the response object and get the choices part.

The API stores answers inside something called choices.

Why “choices”? Because the API can technically return more than one answer if requested. For example, you could ask for multiple possible completions.

So choices is a list.

Conceptually, it is like this:

choices = [
    first_answer,
    second_answer,
    third_answer
]

In this code, you only requested one answer, so usually the list has only one item.

`Why [0]?`

Python lists start counting at 0, not 1.

So:

`choices[0]`

means:

Give me the first answer in the choices list.

If there were more answers, they would be:

`choices[0]  # first answer`
`choices[1]  # second answer`
`choices[2]  # third answer`

In this script, there is only one answer, so we use:

`response.choices[0].message`

After selecting the first choice, this part:

`.message`

gets the message object inside that choice.

That message object has information like:

role: assistant
content: actual answer text
.content

Finally:

`.content`

gets the actual text written by the AI.

So:

response.choices[0].message.content

means:

From the response, get the list of possible answers, take the first answer, get its message, then get the text content of that message.

```python
print("\n--- Simple math answer ---")
print(response.choices[0].message.content)
```

### SECTION 6: Ask the AI to create a challenging question
Here, we ask the model to create a hard question.

Notice that the prompt says "Respond only with the question."

That is an instruction to reduce extra explanation.

```python
question_prompt = (
    "Please propose a hard, challenging question to assess someone's IQ. "
    "Respond only with the question."
)

messages = [
    {
        "role": "user",
        "content": question_prompt
    }
]
```

The next section uses gpt-4.1-mini.

It is more capable than gpt-4.1-nano, but still relatively lightweight.

```python
response = openai.chat.completions.create(
    model="gpt-4.1-mini",
    messages=messages
)
```

Then it stores the generated question in a variable named question.

```python
question = response.choices[0].message.content

print("\n--- AI-generated question ---")
print(question)
```

### SECTION 7: Ask the AI to answer its own generated question

Now we take the question generated by the AI and send it back as a new prompt.

This is an important agentic idea: the output of one AI call becomes the input for another AI call.

```python
messages = [
    {
        "role": "user",
        "content": question
    }
]

response = openai.chat.completions.create(
    model="gpt-4.1-mini",
    messages=messages
)
```

### Finally, we save the answer in a variable.

```python
answer = response.choices[0].message.content

print("\n--- Answer to AI-generated question ---")
print(answer)
```
